```python
import pandas as pd
import numpy as np
from rdkit import Chem
from rdkit.Chem import Descriptors
from sklearn.model_selection import train_test_split, cross_val_score, GridSearchCV
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score, roc_auc_score, roc_curve, confusion_matrix
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LogisticRegression
from sklearn.ensemble import RandomForestClassifier
from xgboost import XGBClassifier # Make sure to install: pip install xgboost
import matplotlib.pyplot as plt
import seaborn as sns
import joblib # For saving/loading models

print("Libraries imported successfully!")

# --- Configuration ---
FILE_PATH = 'BBB.csv' # Ensure this file is in the same directory as your script

# --- 1. Load the dataset ---
try:
    df = pd.read_csv(FILE_PATH)
    print(f"\nDataset '{FILE_PATH}' loaded successfully. Shape: {df.shape}")
    print("First 5 rows of the dataset:")
    print(df.head())
    print("\nDataset information:")
    df.info()
except FileNotFoundError:
    print(f"Error: The file '{FILE_PATH}' was not found. Please ensure it's in the correct directory.")
    exit()
except Exception as e:
    print(f"An error occurred while loading the dataset: {e}")
    exit()

# Drop any "useless" columns (assuming 'Compounds' and 'BBB-Class' are the only relevant ones)
print("\nOriginal columns:", df.columns.tolist())
if 'Compounds' not in df.columns or 'BBB-Class' not in df.columns:
    print("Error: 'Compounds' or 'BBB-Class' column not found. Please check column names.")
    print("Available columns:", df.columns.tolist())
    exit()

# --- 2. SMILES to Molecular Descriptors (Featurization) ---

# Get a list of all RDKit descriptor functions
descriptor_names = [desc[0] for desc in Descriptors.descList]

# Function to generate RDKit descriptors from SMILES
# Modified to return (descriptors, True) if valid, or (None, False) if problematic
def generate_rdkit_descriptors(smiles):
    mol = Chem.MolFromSmiles(smiles)
    if mol is None:
        return None, False # Indicate problematic
    
    descriptors = []
    for desc_name, desc_func in Descriptors.descList:
        try:
            descriptors.append(desc_func(mol))
        except Exception:
            descriptors.append(np.nan) # Append NaN if calculation fails
    return descriptors, True # Indicate valid and return descriptors

print("\nGenerating RDKit molecular descriptors from SMILES strings...")

all_descriptors_data = []
valid_indices = []
problematic_smiles_list = [] # List to store problematic SMILES for logging

for i, smiles_str in enumerate(df['Compounds']):
    if pd.isna(smiles_str) or not isinstance(smiles_str, str):
        problematic_smiles_list.append({'original_index': i, 'SMILES': smiles_str, 'reason': "NaN or Not a valid string type"})
        continue

    desc_values, is_valid = generate_rdkit_descriptors(smiles_str)
    if not is_valid:
        problematic_smiles_list.append({'original_index': i, 'SMILES': smiles_str, 'reason': "RDKit parsing error"})
    else:
        all_descriptors_data.append(desc_values)
        valid_indices.append(i)

if not all_descriptors_data:
    print("Error: No valid descriptors could be generated. Check SMILES column for issues.")
    exit()

# Create a DataFrame for the descriptors
descriptors_df = pd.DataFrame(all_descriptors_data, columns=descriptor_names)

# Re-align with the original DataFrame (only valid SMILES rows)
df_processed = df.iloc[valid_indices].copy().reset_index(drop=True)
df_processed = pd.concat([df_processed, descriptors_df], axis=1)

print(f"\nDataFrame after descriptor generation. Shape: {df_processed.shape}")
print("First 5 rows of the processed DataFrame:")
print(df_processed.head())

# --- Save Problematic SMILES (Training Data) ---
if problematic_smiles_list:
    problematic_df = pd.DataFrame(problematic_smiles_list)
    output_filename = 'problematic_smiles_training.csv'
    problematic_df.to_csv(output_filename, index=False)
    print(f"\nProblematic SMILES from training data saved to '{output_filename}'")
    print(f"Number of problematic SMILES: {len(problematic_smiles_list)}")
else:
    print("\nNo problematic SMILES found in training data.")


# --- 3. Handle Missing Values in Descriptors ---
# Fill NaN values in descriptor columns.
print("\nChecking for and handling NaN values in descriptor columns...")
nan_cols_before = df_processed.columns[df_processed.isnull().any()].tolist()
if nan_cols_before:
    print(f"Columns with NaNs before handling: {nan_cols_before}")
    for col in nan_cols_before:
        if df_processed[col].dtype in ['float64', 'int64']:
            df_processed[col].fillna(df_processed[col].mean(), inplace=True)
        else:
            df_processed[col].fillna(df_processed[col].mode()[0], inplace=True)
    print("NaN values filled using column means/modes.")
else:
    print("No NaN values found in descriptor columns.")

nan_cols_after = df_processed.columns[df_processed.isnull().any()].tolist()
if not nan_cols_after:
    print("All NaN values successfully handled.")
else:
    print(f"Warning: NaNs still present in columns: {nan_cols_after}")


# --- 4. Data Preparation: Features (X) and Target (y) ---
X = df_processed[descriptor_names]
y = df_processed['BBB-Class']

print(f"\nFeatures (X) shape: {X.shape}")
print(f"Target (y) shape: {y.shape}")
print("Target variable value counts:")
print(y.value_counts()) # Check for class imbalance

# Scale numerical features
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)
X_scaled_df = pd.DataFrame(X_scaled, columns=descriptor_names)
print("\nFeatures scaled using StandardScaler.")

# --- 5. Data Splitting ---
X_train, X_test, y_train, y_test = train_test_split(X_scaled_df, y, test_size=0.2, random_state=42, stratify=y)

print(f"\nTrain set shape: X_train {X_train.shape}, y_train {y_train.shape}")
print(f"Test set shape: X_test {X_test.shape}, y_test {y_test.shape}")

# --- 6. Model Training and Evaluation ---

def train_and_evaluate_model(model, model_name, X_train, y_train, X_test, y_test):
    print(f"\n--- Training {model_name} ---")
    model.fit(X_train, y_train)
    y_pred = model.predict(X_test)
    y_prob = model.predict_proba(X_test)[:, 1] if hasattr(model, 'predict_proba') else None

    accuracy = accuracy_score(y_test, y_pred)
    precision = precision_score(y_test, y_pred)
    recall = recall_score(y_test, y_pred)
    f1 = f1_score(y_test, y_pred)
    roc_auc = roc_auc_score(y_test, y_prob) if y_prob is not None else 'N/A'
    cm = confusion_matrix(y_test, y_pred)

    print(f"{model_name} Metrics:")
    print(f"Accuracy: {accuracy:.4f}")
    print(f"Precision: {precision:.4f}")
    print(f"Recall: {recall:.4f}")
    print(f"F1-Score: {f1:.4f}")
    print(f"ROC AUC: {roc_auc:.4f}" if y_prob is not None else "ROC AUC: N/A (model does not have predict_proba)")
    print("Confusion Matrix:\n", cm)

    if y_prob is not None:
        fpr, tpr, _ = roc_curve(y_test, y_prob)
        plt.figure(figsize=(6, 5))
        plt.plot(fpr, tpr, color='darkorange', lw=2, label=f'ROC curve (area = {roc_auc:.2f})')
        plt.plot([0, 1], [0, 1], color='navy', lw=2, linestyle='--')
        plt.xlim([0.0, 1.0])
        plt.ylim([0.0, 1.05])
        plt.xlabel('False Positive Rate')
        plt.ylabel('True Positive Rate')
        plt.title(f'Receiver Operating Characteristic - {model_name}')
        plt.legend(loc="lower right")
        plt.show()

    return model, {'accuracy': accuracy, 'precision': precision, 'recall': recall, 'f1': f1, 'roc_auc': roc_auc}

models = {
    "Logistic Regression": LogisticRegression(random_state=42, solver='liblinear'),
    "Random Forest": RandomForestClassifier(random_state=42),
    "XGBoost": XGBClassifier(random_state=42, use_label_encoder=False, eval_metric='logloss')
}

trained_models = {}
evaluation_results = {}

for name, model in models.items():
    trained_model, metrics = train_and_evaluate_model(model, name, X_train, y_train, X_test, y_test)
    trained_models[name] = trained_model
    evaluation_results[name] = metrics

print("\n--- Summary of Model Performance ---")
for model_name, metrics in evaluation_results.items():
    print(f"\n{model_name}:")
    for metric, value in metrics.items():
        print(f"  {metric.replace('_', ' ').title()}: {value:.4f}" if isinstance(value, float) else f"  {metric.replace('_', ' ').title()}: {value}")


# --- 7. Hyperparameter Tuning (Example with Random Forest) ---
print("\n--- Hyperparameter Tuning Example (Random Forest) ---")
param_grid = {
    'n_estimators': [100, 200, 300],
    'max_depth': [None, 10, 20],
    'min_samples_split': [2, 5],
    'min_samples_leaf': [1, 2]
}

rf_clf = RandomForestClassifier(random_state=42)
grid_search = GridSearchCV(estimator=rf_clf, param_grid=param_grid, cv=5, scoring='roc_auc', n_jobs=-1, verbose=1)

print("Starting GridSearchCV for Random Forest...")
grid_search.fit(X_train, y_train)

print(f"\nBest parameters for Random Forest: {grid_search.best_params_}")
print(f"Best ROC AUC score on training data (cross-validation): {grid_search.best_score_:.4f}")

best_rf_model = grid_search.best_estimator_
_, best_rf_metrics = train_and_evaluate_model(best_rf_model, "Tuned Random Forest", X_train, y_train, X_test, y_test)
evaluation_results["Tuned Random Forest"] = best_rf_metrics


# --- 8. Model Saving ---
# You can save the best performing model for future use
best_model_name = "Tuned Random Forest"
final_model = trained_models[best_model_name] if best_model_name not in evaluation_results else best_rf_model
model_filename = f'{best_model_name.replace(" ", "_").lower()}_bbb_prediction_model.pkl'

try:
    joblib.dump(final_model, model_filename)
    print(f"\nBest model ({best_model_name}) saved to {model_filename}")
except Exception as e:
    print(f"Error saving model: {e}")

print("\n--- Model Training and Prediction Pipeline Complete ---")
print("You can now use the saved model for predictions on new SMILES strings.")
```

```python
import pandas as pd
import numpy as np
from rdkit import Chem
from rdkit.Chem import Descriptors
from sklearn.preprocessing import StandardScaler
import joblib # For loading models and potentially the scaler

print("Libraries imported successfully!")

# --- Configuration ---
# Make sure these file paths are correct
ORIGINAL_TRAINING_DATA_PATH = 'BBB.csv'
EXTERNAL_DATA_PATH = 'External_set.csv'
SAVED_MODEL_PATH = 'tuned_random_forest_bbb_prediction_model.pkl' # !!! IMPORTANT: Update this if your model file has a different name !!!

# --- 1. Define the RDKit Descriptor Generation Function ---
# This function must be identical to the one used during training
descriptor_names = [desc[0] for desc in Descriptors.descList]

def generate_rdkit_descriptors(smiles):
    mol = Chem.MolFromSmiles(smiles)
    if mol is None:
        return None, False # Indicate problematic
    
    descriptors = []
    for desc_name, desc_func in Descriptors.descList:
        try:
            descriptors.append(desc_func(mol))
        except Exception:
            descriptors.append(np.nan) # Append NaN if calculation fails
    return descriptors, True # Indicate valid and return descriptors

print("\nDescriptor generation function defined.")

# --- 2. Load and preprocess the ORIGINAL training data to fit scaler and get imputation means ---
# This step is crucial to ensure consistent preprocessing with the training phase.
print(f"\nLoading original training data from '{ORIGINAL_TRAINING_DATA_PATH}' to fit preprocessors...")
try:
    df_train_original = pd.read_csv(ORIGINAL_TRAINING_DATA_PATH)
except FileNotFoundError:
    print(f"Error: Original training data '{ORIGINAL_TRAINING_DATA_PATH}' not found. Cannot set up preprocessors.")
    exit()

print("Generating descriptors for original training data to fit scaler and get imputation means...")
all_descriptors_data_train = []
valid_indices_train = []
# Note: We are NOT collecting problematic SMILES from original training data here again,
# as that's handled by the training script. This section is purely for preprocessor fitting.
for i, smiles_str in enumerate(df_train_original['Compounds']):
    if pd.isna(smiles_str) or not isinstance(smiles_str, str):
        continue # Skip for preprocessor fitting
    desc_values, is_valid = generate_rdkit_descriptors(smiles_str)
    if is_valid:
        all_descriptors_data_train.append(desc_values)
        valid_indices_train.append(i)

if not all_descriptors_data_train:
    print("Error: No valid descriptors could be generated from original training data for preprocessor fitting.")
    exit()

descriptors_df_train = pd.DataFrame(all_descriptors_data_train, columns=descriptor_names)

# Store means for imputation (from training data)
imputation_means = descriptors_df_train.mean()
descriptors_df_train.fillna(imputation_means, inplace=True) # Apply imputation to training data too for fitting scaler

# Fit the StandardScaler on the ORIGINAL training descriptors
scaler = StandardScaler()
scaler.fit(descriptors_df_train)
print("StandardScaler fitted on original training data.")
print("Imputation means calculated from original training data.")

# --- 3. Load the External Dataset ---
print(f"\nLoading external dataset from '{EXTERNAL_DATA_PATH}'...")
try:
    df_external = pd.read_csv(EXTERNAL_DATA_PATH)
    print("External dataset loaded successfully. Head:")
    print(df_external.head())
    print("\nExternal dataset info:")
    df_external.info()
except FileNotFoundError:
    print(f"Error: External dataset '{EXTERNAL_DATA_PATH}' not found. Please check the file path.")
    exit()
except Exception as e:
    print(f"An error occurred while loading the external dataset: {e}")
    exit()

# Ensure 'Compounds' column exists
if 'SMILES' not in df_external.columns:
    print("Error: 'SMILES' column not found in the external dataset. Please ensure the column name is correct.")
    print("Available columns:", df_external.columns.tolist())
    exit()

# --- 4. Featurize the External Data ---
print("\nGenerating RDKit molecular descriptors for the external dataset...")
all_descriptors_data_external = []
original_external_indices = [] # To map predictions back to original external df indices
problematic_smiles_external_list = [] # List to store problematic SMILES from external data

for i, smiles_str in enumerate(df_external['SMILES']):
    if pd.isna(smiles_str) or not isinstance(smiles_str, str):
        problematic_smiles_external_list.append({'original_index': i, 'SMILES': smiles_str, 'reason': "NaN or Not a valid string type"})
        continue

    desc_values, is_valid = generate_rdkit_descriptors(smiles_str)
    if not is_valid:
        problematic_smiles_external_list.append({'original_index': i, 'SMILES': smiles_str, 'reason': "RDKit parsing error"})
    else:
        all_descriptors_data_external.append(desc_values)
        original_external_indices.append(i)

if not all_descriptors_data_external:
    print("Warning: No valid descriptors could be generated from the external dataset. No predictions will be made.")
    # Save problematic SMILES even if no valid ones
    if problematic_smiles_external_list:
        problematic_external_df = pd.DataFrame(problematic_smiles_external_list)
        output_filename_external = 'problematic_smiles_external.csv'
        problematic_external_df.to_csv(output_filename_external, index=False)
        print(f"\nProblematic SMILES from external data saved to '{output_filename_external}'")
        print(f"Number of problematic SMILES: {len(problematic_smiles_external_list)}")
    exit() # Exit since no valid data to predict on

X_external_raw = pd.DataFrame(all_descriptors_data_external, columns=descriptor_names)

# --- 5. Preprocess the External Data (Imputation and Scaling) ---
print("\nApplying imputation (using training data means) and scaling to external data...")

# Impute NaNs using means from the training data
X_external_imputed = X_external_raw.fillna(imputation_means)

# Scale features using the StandardScaler fitted on training data
X_external_scaled = scaler.transform(X_external_imputed)
X_external_processed = pd.DataFrame(X_external_scaled, columns=descriptor_names)

print("External data successfully featurized and preprocessed.")
print(f"Processed external features shape: {X_external_processed.shape}")

# --- Save Problematic SMILES (External Data) ---
if problematic_smiles_external_list:
    problematic_external_df = pd.DataFrame(problematic_smiles_external_list)
    output_filename_external = 'problematic_smiles_external.csv'
    problematic_external_df.to_csv(output_filename_external, index=False)
    print(f"\nProblematic SMILES from external data saved to '{output_filename_external}'")
    print(f"Number of problematic SMILES: {len(problematic_smiles_external_list)}")
else:
    print("\nNo problematic SMILES found in external data.")

# --- 6. Load the Trained Model ---
print(f"\nLoading the trained model from '{SAVED_MODEL_PATH}'...")
try:
    model = joblib.load(SAVED_MODEL_PATH)
    print("Model loaded successfully.")
    print(f"Model type: {type(model).__name__}")
except FileNotFoundError:
    print(f"Error: Model file '{SAVED_MODEL_PATH}' not found. Please ensure the path and filename are correct.")
    print("Did you run the previous script to save a model? If so, what was the exact filename?")
    exit()
except Exception as e:
    print(f"An error occurred while loading the model: {e}")
    exit()

# --- 7. Make Predictions ---
print("\nMaking predictions on the external dataset...")
predictions = model.predict(X_external_processed)
probabilities = None
if hasattr(model, 'predict_proba'):
    probabilities = model.predict_proba(X_external_processed)[:, 1]

print("\nPredictions complete.")

# --- 8. Output Results ---
results_df = df_external.iloc[original_external_indices].copy().reset_index(drop=True)
results_df['Predicted_BBB_Class'] = predictions
if probabilities is not None:
    results_df['Predicted_Permeability_Probability'] = probabilities

print("\n--- Prediction Results for External Set ---")
print(results_df.head(10))

output_predictions_filename = 'External_set_BBB_predictions.csv'
results_df.to_csv(output_predictions_filename, index=False)
print(f"\nPredictions saved to '{output_predictions_filename}'")

print("\n--- Prediction Pipeline for External Set Complete ---")
```
